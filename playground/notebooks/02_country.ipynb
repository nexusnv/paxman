# Country — Canonicalizing country values with Paxman

**Domain:** country  
**Capability contract:** `Country` (frozen `CanonicalCountryContract`)  
**Public API:** `paxman.canonicalize(input_data, contract)`

Follows the **reference template** (`00_email.ipynb`). Run every cell in order; no hidden state. Country canonicalizes names, alpha-2, alpha-3, numeric M49, and synonyms into the ISO 3166-1 **alpha-2** code.

> Run every cell in order. `artifact.value` holds the canonical string when `artifact.status == "CANONICALIZED"`; otherwise it is `None`.

```python
from paxman import canonicalize, Country, ContractError, CanonicalizationError

def show(raw, contract):
    """Canonicalize `raw`; print status + value."""
    artifact = canonicalize(raw, contract)
    if artifact.status.name == "CANONICALIZED":
        print(f"{raw!r:38} -> {artifact.status.name:14} {artifact.value!r}")
    else:
        print(f"{raw!r:38} -> {artifact.status.name:14} (no canonical value)")
    return artifact
```

Many spellings collapse to the same alpha-2 code: synonym (`USA`), case-folded alpha-2 (`us`), name (`United States`), numeric (`840`), and alpha-3 (`DEU`).

```python
show("USA", Country())                  # synonym
show("us", Country())                   # alpha-2
show("United States", Country())        # name
show("840", Country())                  # numeric M49
show("DEU", Country())                  # alpha-3
show("malaysia", Country(allow_name=True))
```

Policy gates which input *kinds* are accepted. Historical names (`Burma`) need `historical_names=True`; an unknown token is `INVALID`.

```python
show("Burma", Country(historical_names=True))  # MM
show("Atlantis", Country())                    # INVALID
show("DEU", Country(allow_alpha3=False))       # INVALID
```

You can add your own replayable aliases via `extra_synonyms`.

```python
show("america", Country(extra_synonyms={"america": "US"}))
```
