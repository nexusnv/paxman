# Geolocation — Canonicalizing geolocation values with Paxman

**Domain:** geolocation  
**Capability contract:** `Geolocation` (frozen `CanonicalGeolocationContract`)  
**Public API:** `paxman.canonicalize(input_data, contract)`

Follows the **reference template** (`00_email.ipynb`). Run every cell in order; no hidden state. Geolocation canonicalizes coordinates into `"<lat>,<lon>"` at a fixed decimal precision. The default requires an explicit hemisphere sign.

> Run every cell in order. `artifact.value` holds the canonical string when `artifact.status == "CANONICALIZED"`; otherwise it is `None`.

```python
from paxman import canonicalize, Geolocation, ContractError, CanonicalizationError

def show(raw, contract):
    """Canonicalize `raw`; print status + value."""
    artifact = canonicalize(raw, contract)
    if artifact.status.name == "CANONICALIZED":
        print(f"{raw!r:38} -> {artifact.status.name:14} {artifact.value!r}")
    else:
        print(f"{raw!r:38} -> {artifact.status.name:14} (no canonical value)")
    return artifact
```

With `require_hemisphere=False`, unsigned axes default to positive. Hemisphere letters (`N`/`S`/`E`/`W`) are accepted when attached to the number.

```python
show("40.7128, -74.0060", Geolocation(require_hemisphere=False))
show("40.7128N 74.0060W", Geolocation())
```

With the default `require_hemisphere=True`, an unsigned axis is `AMBIGUOUS` — Paxman enumerates the sign readings rather than guessing which hemisphere you meant.

```python
show("40.7128, -74.0060", Geolocation())   # AMBIGUOUS
show("40.7128, 74.0060", Geolocation())    # AMBIGUOUS
```

Out-of-range and malformed inputs are `INVALID`. Hemisphere letters must be attached to the number (`40.7128N 74.0060W` works; `N40.7128` does not).

```python
show("91, 0", Geolocation())                 # INVALID
show("40.7128 -74.0060", Geolocation())      # INVALID
show("N40.7128 W74.0060", Geolocation())     # INVALID
```
