# IP — Canonicalizing ip values with Paxman

**Domain:** ip  
**Capability contract:** `IP` (frozen `CanonicalIPContract`)  
**Public API:** `paxman.canonicalize(input_data, contract)`

Follows the **reference template** (`00_email.ipynb`). Run every cell in order; no hidden state. IP canonicalizes addresses to RFC 4291 (IPv4) / RFC 5952 (IPv6, lowercased + compressed).

> Run every cell in order. `artifact.value` holds the canonical string when `artifact.status == "CANONICALIZED"`; otherwise it is `None`.

```python
from paxman import canonicalize, IP, ContractError, CanonicalizationError

def show(raw, contract):
    """Canonicalize `raw`; print status + value."""
    artifact = canonicalize(raw, contract)
    if artifact.status.name == "CANONICALIZED":
        print(f"{raw!r:38} -> {artifact.status.name:14} {artifact.value!r}")
    else:
        print(f"{raw!r:38} -> {artifact.status.name:14} (no canonical value)")
    return artifact
```

IPv4 leading zeros are stripped; IPv6 is lowercased and compressed; a zone id is preserved (lowercased).

```python
show("192.168.001.001", IP())
show("2001:DB8::1", IP())
show("2001:db8::1%eth0", IP())
```

Policy gates address families. `allow_ipv4=False` + an IPv4 input is `INVALID` (the capability declines that family); malformed addresses are `INVALID`, never coerced.

```python
show("999.1.1.1", IP())                   # INVALID
show("192.168.1.1", IP(allow_ipv4=False)) # INVALID
```
