# URL — Canonicalizing url values with Paxman

**Domain:** url  
**Capability contract:** `URL` (frozen `CanonicalURLContract`)  
**Public API:** `paxman.canonicalize(input_data, contract)`

Follows the **reference template** (`00_email.ipynb`). Run every cell in order; no hidden state. URL normalizes to RFC 3986 §6.2.2 form (scheme/host lowercased, default port elided, dot-segments removed).

> Run every cell in order. `artifact.value` holds the canonical string when `artifact.status == "CANONICALIZED"`; otherwise it is `None`.

```python
from paxman import canonicalize, URL, ContractError

def show(raw, contract):
    """Canonicalize `raw`; print status, value, and evidence."""
    artifact = canonicalize(raw, contract)
    if artifact.status.name == "CANONICALIZED":
        print(f"{raw!r:38} -> {artifact.status.name:14} {artifact.value!r}")
    else:
        print(f"{raw!r:38} -> {artifact.status.name:14} (no canonical value)")
    for e in artifact.evidence:
        print(f"    evidence: {e.rule} {e.detail!r}")
    return artifact
```

Scheme and host are lowercased; the default port (`:80`) is elided; `..` dot-segments are resolved; fragments are dropped by default.

```python
show("HTTP://Example.COM:80/foo/bar", URL())
show("http://example.com/foo/../bar", URL())
show("http://user:pass@example.com/path#frag", URL())
show("https://example.com/a?b=2&c=1", URL(sort_query=True))
```

`scheme_allow` declares accepted schemes. An input whose scheme is NOT allowed is `UNSUPPORTED` — the capability declines, distinct from `INVALID`. Note: the URL capability is lenient — an unparseable string is passed through as `CANONICALIZED`, so do not rely on it to validate free text.

```python
show("ftp://example.com/file", URL())
show("ftp://example.com/file", URL(scheme_allow=("https",)))
show("not a url", URL())
```
